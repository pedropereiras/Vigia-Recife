# 6. Tratamento dos Dados

Após a etapa de limpeza foram definidas as estratégias de tratamento para cada inconsistência identificada.

## Decisões de Tratamento

Com base nas verificações realizadas durante a etapa de limpeza, foram adotadas as seguintes decisões:

- Datas inválidas são convertidas para `NaT`, preservando o registro original e permitindo que análises temporais ignorem apenas os casos inconsistentes.

- Idades inferiores a 0 anos ou superiores a 110 anos são consideradas inválidas e substituídas por valores ausentes (`NaN`).

- Registros cujo horário corresponde exatamente a **21:00:00** são tratados como **horário estimado** (valor padrão do sistema). Esses registros permanecem na base principal, porém recebem uma coluna auxiliar (`hora_valida`) para evitar que análises temporais utilizem horários potencialmente artificiais.

- Categorias `"NÃO IDENTIFICADO"` em `race` e `genre` são mantidas na base (não removidas), mas destacadas separadamente nas análises demográficas, já que representam subnotificação e não um grupo demográfico real.

```python
def tratar_datas(df, coluna_data="data"):
    """
    Converte a coluna de datas para datetime.
    Datas inválidas tornam-se NaT.
    """

    df = df.copy()

    df[coluna_data] = pd.to_datetime(
        df[coluna_data],
        format="%d/%m/%Y, %H:%M:%S",
        errors="coerce"
    )

    return df


def tratar_idades(df, coluna="age"):
    """
    Idades menores que 0 ou maiores que 110
    são consideradas inválidas.
    """

    df = df.copy()

    df.loc[
        (df[coluna] < 0) |
        (df[coluna] > 110),
        coluna
    ] = np.nan

    return df


def tratar_hora_estimada(df, coluna_data="data"):
    """
    Marca registros cujo horário é o valor padrão do sistema
    (21:00:00) e cria uma coluna de hora 'válida' (NaN nesses
    casos) para uso em análises que dependem da hora exata.
    """

    df = df.copy()

    df["hora_estimada"] = (

        (df[coluna_data].dt.hour == 21) &
        (df[coluna_data].dt.minute == 0) &
        (df[coluna_data].dt.second == 0)

    )

    df["hora_valida"] = (
        df[coluna_data]
        .dt.hour
        .astype("float")
    )

    df.loc[
        df["hora_estimada"],
        "hora_valida"
    ] = np.nan

    return df


def relatorio_nulos(df):
    """
    Exibe relatório de valores ausentes.
    """

    relatorio = (

        df.isnull()

        .sum()

        .sort_values(ascending=False)

        .to_frame("Valores Ausentes")

    )

    relatorio["Percentual (%)"] = (

        relatorio["Valores Ausentes"]

        / len(df)

        * 100

    ).round(2)

    return relatorio
```

```python
#aplicando o tratamento
df_limpo = tratar_datas(df_limpo)

df_limpo = tratar_idades(df_limpo)

df_limpo = tratar_hora_estimada(df_limpo)

print(
    f"Registros com hora estimada "
    f"(excluídos da análise de horário): "
    f"{df_limpo['hora_estimada'].sum()}"
)

relatorio_nulos(df_limpo).head(15)
```

# 7. Engenharia de Atributos

Nesta etapa são criadas novas variáveis derivadas da coluna de data e hora.

Essas variáveis facilitam análises temporais e também servem como atributos na clusterização mais adiante.

As variáveis derivadas de **hora do dia** (`hora`, `periodo`) usam a coluna `hora_valida`, que já exclui os registros com horário padrão identificados na Seção 5.

```python
def criar_features_temporais(df):

    df = df.copy()

    month_map = {

        1: "JANEIRO",
        2: "FEVEREIRO",
        3: "MARÇO",
        4: "ABRIL",
        5: "MAIO",
        6: "JUNHO",
        7: "JULHO",
        8: "AGOSTO",
        9: "SETEMBRO",
        10: "OUTUBRO",
        11: "NOVEMBRO",
        12: "DEZEMBRO"

    }


    day_map = {

        "Monday": "SEGUNDA-FEIRA",
        "Tuesday": "TERÇA-FEIRA",
        "Wednesday": "QUARTA-FEIRA",
        "Thursday": "QUINTA-FEIRA",
        "Friday": "SEXTA-FEIRA",
        "Saturday": "SÁBADO",
        "Sunday": "DOMINGO"

    }


    df["ano"] = df["data"].dt.year

    df["mes"] = df["data"].dt.month

    df["nome_mes"] = df["data"].dt.month.map(month_map)

    df["dia"] = df["data"].dt.day

    df["dia_semana"] = (

        df["data"]

        .dt.day_name()

        .map(day_map)

    )

    df["trimestre"] = (

        df["data"]

        .dt.quarter

    )

    df["fim_semana"] = (

        df["dia_semana"]

        .isin(

            [

                "SÁBADO",

                "DOMINGO"

            ]

        )

    )


    df["hora"] = df["hora_valida"]

    def periodo(hora):

        if pd.isna(hora):

            return np.nan

        if hora < 6:

            return "MADRUGADA"

        elif hora < 12:

            return "MANHÃ"

        elif hora < 18:

            return "TARDE"

        else:

            return "NOITE"

    df["periodo"] = (

        df["hora"]

        .apply(periodo)

    )

    return df
```

```python

df_analise = criar_features_temporais(df_limpo)
```

## 9.1 Distribuição das Ocorrências por Bairro

```python
def grafico_bairro(df, top=10):

    bairros = (

        df["neighborhood"]

        .value_counts()

        .head(top)

        .sort_values()

    )

    fig, ax = plt.subplots(figsize=(12,7))

    bars = ax.barh(

        bairros.index,

        bairros.values,

        color="#D32F2F"

    )

    ax.set_title(

        f"Top {top} bairros com maior número de ocorrências",

        fontsize=16,

        weight="bold"

    )

    ax.set_xlabel("Quantidade de ocorrências")

    for barra in bars:

        largura = barra.get_width()

        ax.text(

            largura + 2,

            barra.get_y() + barra.get_height()/2,

            f"{int(largura)}",

            va="center"

        )

    plt.tight_layout()

    plt.show()
```

## 9.3 Perfil Demográfico das Vítimas

Nesta seção são analisadas características demográficas das vítimas registradas:
gênero, raça e faixa etária.

> **Atenção — qualidade dos dados:** a categoria `"NÃO IDENTIFICADO"` corresponde a **72% dos registros de raça** e a uma parcela de `genre`. Isso não deve ser interpretado como um "grupo racial predominante" — é subnotificação. Nas leituras abaixo, tratamos essa categoria separadamente para não distorcer a interpretação.

```python
def grafico_genero(df):
    dados = df["genre"].value_counts()

    fig, ax = plt.subplots(figsize=(9, 5))

    bars = ax.bar(
        dados.index,
        dados.values,
        color="#1976D2"
    )

    total = dados.sum()

    for barra in bars:
        valor = barra.get_height()
        ax.text(
            barra.get_x() + barra.get_width() / 2,
            valor,
            f"{valor}\n({valor/total:.1%})",
            ha="center"
        )

    plt.title("Distribuição por Gênero", fontsize=15, weight="bold")
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()

grafico_genero(df_analise)
```

### Insight

A distribuição por gênero mostra forte predominância de vítimas do sexo masculino cis, padrão consistente com estatísticas nacionais de violência letal no Brasil, e permite direcionar políticas públicas específicas.

```python
def grafico_raca(df):
    dados = df["race"].fillna("NÃO INFORMADO").value_counts()

    cores = [
        "#9E9E9E" if cat == "NÃO IDENTIFICADO" else "#388E3C"
        for cat in dados.index
    ]

    fig, ax = plt.subplots(figsize=(9, 6))

    bars = ax.barh(
        dados.index,
        dados.values,
        color=cores
    )

    total = dados.sum()

    for barra in bars:
        largura = barra.get_width()
        ax.text(
            largura + 2,
            barra.get_y() + barra.get_height() / 2,
            f"{largura} ({largura/total:.1%})",
            va="center"
        )

    plt.title(
        "Distribuição por Raça\n(cinza = não identificado / subnotificado)",
        fontsize=14,
        weight="bold"
    )

    plt.tight_layout()
    plt.show()

grafico_raca(df_analise)
```

## 10.4 Clusterização de Bairros

**Melhoria em relação à versão anterior:** o agrupamento original usava apenas 3 variáveis agregadas por bairro (contagem de ocorrências, idade média e % de ação policial), o que fazia os clusters ficarem dominados pelo volume total de ocorrências — dois bairros com perfis de violência muito diferentes, mas volume parecido, caíam no mesmo grupo.

Aqui o perfil de cada bairro é enriquecido com:

- proporção de ocorrências por período do dia (madrugada/manhã/tarde/noite);
- proporção dos principais tipos de ocorrência;
- idade média das vítimas;
- percentual de ação policial;
- percentual de casos em fins de semana.

Isso faz os clusters refletirem **padrões de comportamento**, não apenas volume.

```python
# Base agregada por bairro

agg_volume = df_analise.groupby("neighborhood").agg(
    ocorrencias=("id", "count"),
    idade_media=("age", "mean"),
    acao_policial=("police_action", "mean"),
    pct_fim_semana=("fim_semana", "mean"),
)

# Proporção por período do dia

prop_periodo = (
    pd.crosstab(
        df_analise["neighborhood"],
        df_analise["periodo"],
        normalize="index"
    )
    .add_prefix("periodo_")
)

# Proporção dos 5 tipos de ocorrência mais comuns
# (demais agrupados em "OUTROS")

top_tipos = df_analise["main_reason"].value_counts().head(5).index

df_tipo = df_analise.copy()

df_tipo["main_reason_agrupado"] = df_tipo["main_reason"].where(
    df_tipo["main_reason"].isin(top_tipos),
    "OUTROS"
)

prop_tipo = (
    pd.crosstab(
        df_tipo["neighborhood"],
        df_tipo["main_reason_agrupado"],
        normalize="index"
    )
    .add_prefix("tipo_")
)

cluster = (
    agg_volume
    .join(prop_periodo, how="left")
    .join(prop_tipo, how="left")
)

# Mantém apenas bairros com volume mínimo de ocorrências,
# para evitar que bairros com 1–2 registros distorçam o agrupamento.

MIN_OCORRENCIAS = 10

cluster = cluster[
    cluster["ocorrencias"] >= MIN_OCORRENCIAS
]

cluster = cluster.fillna(
    cluster.mean(numeric_only=True)
)

print(f"Bairros incluídos na clusterização: {len(cluster)}")

cluster.head()
```

```python
K_ESCOLHIDO = 4

scaler = StandardScaler()
X = scaler.fit_transform(cluster)

modelo = KMeans(
    n_clusters=K_ESCOLHIDO,
    random_state=42,
    n_init=10
)

cluster["cluster"] = modelo.fit_predict(X)

cluster[
    [
        "ocorrencias",
        "idade_media",
        "acao_policial",
        "cluster"
    ]
].sort_values("cluster").head(10)
```

```python
pca = PCA(n_components=2)

componentes = pca.fit_transform(X)

plt.figure(figsize=(9, 7))

scatter = plt.scatter(
    componentes[:, 0],
    componentes[:, 1],
    c=cluster["cluster"],
    cmap="Set2",
    s=90,
    edgecolor="black",
    linewidth=.5
)

for i, bairro in enumerate(cluster.index):
    plt.text(
        componentes[i, 0],
        componentes[i, 1],
        bairro,
        fontsize=7
    )

plt.title(
    f"Clusters de bairros (PCA) — variância explicada: {pca.explained_variance_ratio_.sum():.1%}"
)

plt.xlabel("Componente 1")
plt.ylabel("Componente 2")

plt.show()
```

```python
# Perfil médio de cada cluster
# Ajuda a nomear e interpretar os grupos.

resumo_clusters = (
    cluster.groupby("cluster")[
        [
            "ocorrencias",
            "idade_media",
            "acao_policial",
            "pct_fim_semana"
        ]
    ]
    .mean()
    .round(2)
)

resumo_clusters
```

### Insight

Com atributos comportamentais (período do dia, tipo de ocorrência, ação policial), os clusters deixam de refletir apenas "bairros com mais ou menos ocorrências" e passam a distinguir **perfis de violência**: por exemplo, bairros com maior concentração em madrugada versus bairros com maior concentração em horário comercial, ou bairros com maior/menor resposta policial proporcional.

Isso é mais acionável para desenho de políticas públicas segmentadas do que um ranking simples por volume.

```python

# 4. Limpeza
df_limpo = df.copy()

def padronizar_texto(df):
    dados = df.copy()
    colunas = dados.select_dtypes(include="object").columns
    for coluna in colunas:
        dados[coluna] = dados[coluna].astype(str).str.strip().str.upper()
    return dados

df_limpo = padronizar_texto(df_limpo)
dup = df_limpo.duplicated().sum()
df_limpo = df_limpo.drop_duplicates()

# 6. Tratamento
def tratar_datas(df, coluna_data="data"):
    df = df.copy()
    df[coluna_data] = pd.to_datetime(df[coluna_data], format="%d/%m/%Y, %H:%M:%S", errors="coerce")
    return df

def tratar_idades(df, coluna="age"):
    df = df.copy()
    df.loc[(df[coluna] < 0) | (df[coluna] > 110), coluna] = np.nan
    return df

def tratar_hora_estimada(df, coluna_data="data"):
    df = df.copy()
    df["hora_estimada"] = ((df[coluna_data].dt.hour == 21) & (df[coluna_data].dt.minute == 0) & (df[coluna_data].dt.second == 0))
    df["hora_valida"] = df[coluna_data].dt.hour.astype("float")
    df.loc[df["hora_estimada"], "hora_valida"] = np.nan
    return df

df_limpo = tratar_datas(df_limpo)
df_limpo = tratar_idades(df_limpo)
df_limpo = tratar_hora_estimada(df_limpo)

# 7. Engenharia de atributos
def criar_features_temporais(df):
    df = df.copy()
    day_map = {"Monday":"SEGUNDA-FEIRA","Tuesday":"TERÇA-FEIRA","Wednesday":"QUARTA-FEIRA","Thursday":"QUINTA-FEIRA","Friday":"SEXTA-FEIRA","Saturday":"SÁBADO","Sunday":"DOMINGO"}
    df["ano"] = df["data"].dt.year
    df["mes"] = df["data"].dt.month
    df["dia_semana"] = df["data"].dt.day_name().map(day_map)
    df["fim_semana"] = df["dia_semana"].isin(["SÁBADO","DOMINGO"])
    df["hora"] = df["hora_valida"]
    def periodo(hora):
        if pd.isna(hora): return np.nan
        if hora < 6: return "MADRUGADA"
        elif hora < 12: return "MANHÃ"
        elif hora < 18: return "TARDE"
        else: return "NOITE"
    df["periodo"] = df["hora"].apply(periodo)
    return df

df_analise = criar_features_temporais(df_limpo)

report = []
report.append(f"REGISTROS BRUTOS: {df.shape[0]}")
report.append(f"DUPLICADOS REMOVIDOS: {dup}")
report.append(f"REGISTROS APÓS LIMPEZA: {df_limpo.shape[0]}")
report.append(f"BAIRROS ÚNICOS: {df_limpo['neighborhood'].nunique()}")
report.append(f"PERÍODO: {df_limpo['data'].min()} até {df_limpo['data'].max()}")
report.append(f"REGISTROS COM HORA ESTIMADA (21h artefato): {df_limpo['hora_estimada'].sum()} ({df_limpo['hora_estimada'].mean():.1%})")

# 9.1 Top bairros
top_bairros = df_analise["neighborhood"].value_counts().head(15)
report.append("\n=== TOP 15 BAIRROS COM MAIS OCORRÊNCIAS ===")
report.append(top_bairros.to_string())

# 9.2 Série temporal por ano
serie_anual = df_analise.groupby("ano").size()
report.append("\n=== OCORRÊNCIAS POR ANO ===")
report.append(serie_anual.to_string())

# 9.3 Demografia
genero = df_analise["genre"].value_counts()
report.append("\n=== DISTRIBUIÇÃO POR GÊNERO ===")
report.append(genero.to_string())

raca_pct_nao_id = (df_analise["race"] == "NÃO IDENTIFICADO").mean()
report.append(f"\n% RAÇA NÃO IDENTIFICADA: {raca_pct_nao_id:.1%}")
raca_identificada = df_analise[~df_analise["race"].isin(["NÃO IDENTIFICADO"]) & df_analise["race"].notna()]["race"].value_counts(normalize=True)*100
report.append("=== RAÇA (apenas identificados) ===")
report.append(raca_identificada.round(1).to_string())

# Idade
idade = df_analise["age"].dropna()
report.append(f"\nIDADE MÉDIA: {idade.mean():.1f} | MEDIANA: {idade.median():.1f}")

# Tipo de ocorrência
tipo = df_analise["main_reason"].value_counts()
report.append("\n=== TIPOS DE OCORRÊNCIA ===")
report.append(tipo.to_string())

# Período do dia (só horas válidas)
periodo = df_analise["periodo"].value_counts()
report.append("\n=== PERÍODO DO DIA (excluindo hora artefato) ===")
report.append(periodo.to_string())

# Dia da semana
dia_sem = df_analise["dia_semana"].value_counts().reindex(["SEGUNDA-FEIRA","TERÇA-FEIRA","QUARTA-FEIRA","QUINTA-FEIRA","SEXTA-FEIRA","SÁBADO","DOMINGO"])
report.append("\n=== DIA DA SEMANA ===")
report.append(dia_sem.to_string())

# 11. Ação policial
acao_por_bairro = (
    df_analise.groupby("neighborhood")["police_action"]
    .agg(ocorrencias="count", pct_acao="mean")
    .query("ocorrencias >= 20")
    .sort_values("pct_acao", ascending=False)
)
acao_por_bairro["pct_acao"] = (acao_por_bairro["pct_acao"]*100).round(1)
report.append("\n=== TOP 10 BAIRROS COM MAIOR % AÇÃO POLICIAL (min 20 ocorrências) ===")
report.append(acao_por_bairro.head(10).to_string())
report.append("\n=== TOP 10 BAIRROS COM MENOR % AÇÃO POLICIAL (min 20 ocorrências) ===")
report.append(acao_por_bairro.tail(10).sort_values("pct_acao").to_string())

# Tendência estatística
serie_mensal = df_analise.set_index("data").resample("ME").size()
ultimo_mes_completo = pd.Timestamp.today().normalize().replace(day=1) - pd.Timedelta(days=1)
serie_mensal_completa = serie_mensal[serie_mensal.index <= ultimo_mes_completo]
x = np.arange(len(serie_mensal_completa))
resultado = linregress(x, serie_mensal_completa.values)
report.append(f"\n=== TENDÊNCIA TEMPORAL (regressão linear mensal) ===")
report.append(f"Inclinação: {resultado.slope:.3f} | p-valor: {resultado.pvalue:.5f}")
if resultado.pvalue < 0.05:
    direcao = "CRESCIMENTO" if resultado.slope > 0 else "QUEDA"
    report.append(f"Tendência estatisticamente significativa de {direcao}")
else:
    report.append("Sem tendência estatisticamente significativa (p >= 0.05)")

# 10.4 Clusterização
agg_volume = df_analise.groupby("neighborhood").agg(
    ocorrencias=("id","count"),
    idade_media=("age","mean"),
    acao_policial=("police_action","mean"),
    pct_fim_semana=("fim_semana","mean"),
)
prop_periodo = pd.crosstab(df_analise["neighborhood"], df_analise["periodo"], normalize="index").add_prefix("periodo_")
top_tipos = df_analise["main_reason"].value_counts().head(5).index
df_tipo = df_analise.copy()
df_tipo["main_reason_agrupado"] = df_tipo["main_reason"].where(df_tipo["main_reason"].isin(top_tipos), "OUTROS")
prop_tipo = pd.crosstab(df_tipo["neighborhood"], df_tipo["main_reason_agrupado"], normalize="index").add_prefix("tipo_")

cluster = agg_volume.join(prop_periodo, how="left").join(prop_tipo, how="left")
MIN_OCORRENCIAS = 10
cluster = cluster[cluster["ocorrencias"] >= MIN_OCORRENCIAS]
cluster = cluster.fillna(cluster.mean(numeric_only=True))

scaler = StandardScaler()
X = scaler.fit_transform(cluster)

# --- Escolha de k via método do cotovelo + silhueta ---
# (ideia originalmente esboçada no notebook exploratório "vigia_v1"; aqui aplicada
# sobre o X comportamental completo, não apenas lat/long, e levada até o fim:
# escolhe k, roda o KMeans final e atribui o cluster de volta ao dataframe)
K_range = range(2, 9)
inercias = []
silhuetas = []
for k in K_range:
    km_teste = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels_teste = km_teste.fit_predict(X)
    inercias.append(km_teste.inertia_)
    silhuetas.append(silhouette_score(X, labels_teste))

fig, ax1 = plt.subplots(figsize=(9, 5))
ax1.plot(list(K_range), inercias, marker="o", color="#B71C1C", label="Inércia (cotovelo)")
ax1.set_xlabel("Número de clusters (k)")
ax1.set_ylabel("Inércia", color="#B71C1C")
ax2 = ax1.twinx()
ax2.plot(list(K_range), silhuetas, marker="s", color="#1976D2", label="Silhueta")
ax2.set_ylabel("Coeficiente de silhueta", color="#1976D2")
plt.title("Escolha de k — método do cotovelo + silhueta")
fig.tight_layout()
plt.savefig(f"{OUT}/elbow_silhouette.png", dpi=110)
plt.close()

K_ESCOLHIDO = list(K_range)[int(np.argmax(silhuetas))]
report.append(f"\n=== ESCOLHA DE k (cotovelo + silhueta) ===")
report.append(f"Inércias por k: {dict(zip(K_range, [round(i,1) for i in inercias]))}")
report.append(f"Silhuetas por k: {dict(zip(K_range, [round(s,3) for s in silhuetas]))}")
report.append(f"k escolhido (maior silhueta): {K_ESCOLHIDO}")

modelo = KMeans(n_clusters=K_ESCOLHIDO, random_state=42, n_init=10)
cluster["cluster"] = modelo.fit_predict(X)

resumo_clusters = cluster.groupby("cluster")[["ocorrencias","idade_media","acao_policial","pct_fim_semana"]].mean().round(2)
report.append(f"\n=== CLUSTERIZAÇÃO (k={K_ESCOLHIDO}, {len(cluster)} bairros com >= {MIN_OCORRENCIAS} ocorrências) ===")
report.append(resumo_clusters.to_string())

report.append("\n=== BAIRROS POR CLUSTER ===")
for c in sorted(cluster["cluster"].unique()):
    bairros_c = cluster[cluster["cluster"]==c].sort_values("ocorrencias", ascending=False).index.tolist()
    report.append(f"Cluster {c} ({len(bairros_c)} bairros): {', '.join(bairros_c[:12])}{'...' if len(bairros_c)>12 else ''}")

# --- Pirâmide etária por gênero ---
# (ideia originária do notebook exploratório "vigia_v1", corrigida aqui para:
#  usar hora_valida/tratamentos já aplicados, não descartar "não identificado"
#  da base geral — apenas filtra localmente para ESTE gráfico específico — e
#  ficar disponível como função reutilizável por bairro.)
ORDEM_FAIXA = ["CRIANÇA", "ADOLESCENTE", "ADULTO", "IDOSO"]

def montar_piramide(df):
    base = df[
        df["genre"].isin(["HOMEM CIS", "MULHER CIS"])
        & df["ageGroup"].isin(ORDEM_FAIXA)
    ]
    if base.empty:
        return None
    pir = pd.crosstab(base["ageGroup"], base["genre"]).reindex(ORDEM_FAIXA, fill_value=0)
    return pir

def plot_piramide(pir, titulo, caminho):
    pir = pir.copy()
    pir["HOMEM CIS"] = -pir["HOMEM CIS"]
    limite = max(pir["MULHER CIS"].max(), abs(pir["HOMEM CIS"].min())) + 5
    plt.figure(figsize=(10, 6))
    plt.barh(pir.index, pir["HOMEM CIS"], color="#0F4C81", label="Homens")
    plt.barh(pir.index, pir["MULHER CIS"], color="#F28E2B", label="Mulheres")
    plt.axvline(0, color="black", linewidth=1)
    for i, v in enumerate(pir["HOMEM CIS"]):
        if v != 0:
            plt.text(v, i, str(abs(v)), ha="right", va="center", fontsize=9)
    for i, v in enumerate(pir["MULHER CIS"]):
        if v != 0:
            plt.text(v, i, str(v), ha="left", va="center", fontsize=9)
    plt.xlim(-limite, limite)
    ticks = np.linspace(-limite, limite, 7).astype(int)
    plt.xticks(ticks, [abs(t) for t in ticks])
    plt.title(titulo, fontsize=15, weight="bold")
    plt.xlabel("Quantidade de vítimas")
    plt.ylabel("Faixa etária")
    plt.grid(axis="x", linestyle="--", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(caminho, dpi=110)
    plt.close()

pir_geral = montar_piramide(df_analise)
if pir_geral is not None:
    plot_piramide(pir_geral, "Pirâmide Etária das Vítimas — Recife (geral)", f"{OUT}/piramide_geral.png")
    report.append("\n=== PIRÂMIDE ETÁRIA (geral, homem cis vs mulher cis) ===")
    report.append(pir_geral.to_string())

def piramide_bairro(df, bairro, caminho=None):
    dados = df[df["neighborhood"] == bairro]
    pir = montar_piramide(dados)
    if pir is None:
        return None
    if caminho:
        plot_piramide(pir, f"Pirâmide Etária das Vítimas — {bairro}", caminho)
    return pir

# Demonstração da função para o bairro de maior volume
bairro_exemplo = top_bairros.index[0]
pir_bairro = piramide_bairro(df_analise, bairro_exemplo, f"{OUT}/piramide_{bairro_exemplo}.png")
if pir_bairro is not None:
    report.append(f"\n=== PIRÂMIDE ETÁRIA — {bairro_exemplo} (exemplo de uso da função piramide_bairro) ===")
    report.append(pir_bairro.to_string())

# Salvar CSVs úteis para o app
cluster.to_csv(f"{OUT}/bairros_cluster.csv")
top_bairros.to_csv(f"{OUT}/top_bairros.csv")
df_analise.to_csv(f"{OUT}/eventos_tratados.csv", index=False)
acao_por_bairro.to_csv(f"{OUT}/acao_policial_por_bairro.csv")

with open(f"{OUT}/relatorio.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(str(r) for r in report))

print("\n".join(str(r) for r in report))
print("\n\nArquivos salvos em", OUT)
```
